```python
import base64
import jwt
import requests
from cryptography.hazmat.primitives import serialization
from cryptography.hazmat.primitives.asymmetric import rsa
import json

token = 'eyJ0eXAiOiJKV1QiLCJub25jZSI6ImpNWGhLN29HUVhrNFpZTXlXaW52SXZQTFFxcktfX0RKM2xTeHZ3LS1BaGMiLCJhbGciOiJSUzI1NiIsIng1dCI6IlhSdmtvOFA3QTNVYVdTblU3Yk05blQwTWpoQSIsImtpZCI6IlhSdmtvOFA3QTNVYVdTblU3Yk05blQwTWpoQSJ9.eyJhdWQiOiIwMDAwMDAwMy0wMDAwLTAwMDAtYzAwMC0wMDAwMDAwMDAwMDAiLCJpc3MiOiJodHRwczovL3N0cy53aW5kb3dzLm5ldC85Mjg0YjU3ZC1iY2MxLTQzZTMtYjhkNS0zZjQ4MGJmZDRhNzAvIiwiaWF0IjoxNzExMTE2MDY5LCJuYmYiOjE3MTExMTYwNjksImV4cCI6MTcxMTEyMDk3NiwiYWNjdCI6MCwiYWNyIjoiMSIsImFpbyI6IkFZUUFlLzhXQUFBQXU5NEJXL2lndjdDNkduU2lyWFdSb1hjUGw0d25DQWt2Yi9TVk44NmNCTmZ0SGd3Mnhlcno0TnN0TWU5ZVBaK284N2txYzRUejRZU01uUkxJMjJweDYvbkNiRVJPcFJrWERFcU1CQmgzSGc4YlRQcTdiOWtCYS9rS0JCRXljb2NNbThUdDVLWUd2Mm1YQjZoMVFPK01RdXZEYnArbWVvK2pkeFFKSzlTb21rTT0iLCJhbHRzZWNpZCI6IjE6bGl2ZS5jb206MDAwMzAwMDA4MEYwN0JDRCIsImFtciI6WyJwd2QiLCJtZmEiXSwiYXBwX2Rpc3BsYXluYW1lIjoiZW1sYWIuYWkiLCJhcHBpZCI6ImU5NWI4YWZmLWM4OTItNDkyYi1hMzFiLTI2MzI0ZGI0YTc5YiIsImFwcGlkYWNyIjoiMCIsImVtYWlsIjoibWFsYXBoZWV2QGdtYWlsLmNvbSIsImZhbWlseV9uYW1lIjoiTWFsYWZlZXYiLCJnaXZlbl9uYW1lIjoiQWxla3NhbmRyIiwiaWRwIjoibGl2ZS5jb20iLCJpZHR5cCI6InVzZXIiLCJpcGFkZHIiOiIxOTUuMjI0LjE0NS4yNDkiLCJuYW1lIjoiQWxla3NhbmRyIE1hbGFmZWV2Iiwib2lkIjoiZGYwNWQ1MmMtYjgwZC00YTVhLWJkOTMtN2Q4ZWM1M2MzOTE0IiwicGxhdGYiOiI1IiwicHVpZCI6IjEwMDMzRkZGODc2RDBBREYiLCJyaCI6IjAuQVhvQWZiV0Vrc0c4NDBPNDFUOUlDXzFLY0FNQUFBQUFBQUFBd0FBQUFBQUFBQUI2QUZzLiIsInNjcCI6Im9wZW5pZCBwcm9maWxlIFVzZXIuUmVhZCBlbWFpbCIsInNpZ25pbl9zdGF0ZSI6WyJrbXNpIl0sInN1YiI6ImNvRElReHA5STJXUW92RTVsemtkZG9RZk50Z2JpZTFzbFNKd1RoTTRWZDAiLCJ0ZW5hbnRfcmVnaW9uX3Njb3BlIjoiRVUiLCJ0aWQiOiI5Mjg0YjU3ZC1iY2MxLTQzZTMtYjhkNS0zZjQ4MGJmZDRhNzAiLCJ1bmlxdWVfbmFtZSI6ImxpdmUuY29tI21hbGFwaGVldkBnbWFpbC5jb20iLCJ1dGkiOiJlT3ZaZllsdWNrNmxvRGRsM2MxYUFBIiwidmVyIjoiMS4wIiwid2lkcyI6WyI2MmU5MDM5NC02OWY1LTQyMzctOTE5MC0wMTIxNzcxNDVlMTAiLCJiNzlmYmY0ZC0zZWY5LTQ2ODktODE0My03NmIxOTRlODU1MDkiXSwieG1zX3N0Ijp7InN1YiI6IlVscTZDWjJfYkVQNS1KRFZLQVJRa2Q4Q25PWUR3NXlhWHlXZzZ2Y3JfSE0ifSwieG1zX3RjZHQiOjEzODE3MTgzNjEsInhtc190ZGJyIjoiRVUifQ.nUdELJ0yilI13yjXe1CUwP5Nxhtum_fnm-4RDI7HpsONzesUhWa_XKX4BEfqvees93q5r2opGWwLCZHN6AB1gYULlWUikXf--TCSWsRHiqJCULg2ru-GMqRD88KKbgvSQFU_Jht3zJMGf-mtaT4Xg55BbH3i-7r6QeOscE0ywryhuloLRoyeW6pY3W3y3yYDG2w__AIa00ksGjY93KeStI47eIhOhYi6tmVQplD640zGKmDtJG5rhM4MgsKOZ_tZ8oWQcBrzongnCZwNSoOM9Y8u9Kmqqwij5-bmsrZnlMHaVmpIn2n9XVXOj2f0UtUNzKyKZajGUgZStn23B3yn_w'

# Function to retrieve the public key for token validation
def get_public_key(token):
    # Make a request to the OpenID configuration endpoint
    response = requests.get('https://login.microsoftonline.com/common/.well-known/openid-configuration')
    if response.status_code == 200:
        # Extract the jwks_uri from the response
        jwks_uri = response.json()['jwks_uri']
        # Make a request to the jwks_uri to retrieve the public keys
        response = requests.get(jwks_uri)
        if response.status_code == 200:
            # Extract the public keys from the response
            public_keys = response.json()['keys']
            # Find the matching public key based on the token's kid (key ID)
            for key in public_keys:
                if key['kid'] == jwt.get_unverified_header(token)['kid']:
                    return key
    return None

def base64url_to_int(base64url_string):
    # Base64url decode the string
    padding = '=' * (4 - (len(base64url_string) % 4))
    base64url_string += padding
    decoded_string = base64.urlsafe_b64decode(base64url_string)
    # Convert the decoded bytes to an integer
    return int.from_bytes(decoded_string, 'big')

# Function to validate the bearer token
def validate_bearer_token(token):
    # Retrieve the public key for token validation
    public_key = get_public_key(token)

    # Extract the necessary components from the dictionary
    n = base64url_to_int(public_key['n'])
    e = base64url_to_int(public_key['e'])

    # Create a public key object
    public_key_obj = rsa.RSAPublicNumbers(e, n).public_key()

    # Serialize the public key to PEM format
    public_key_pem = public_key_obj.public_bytes(
        encoding=serialization.Encoding.PEM,
        format=serialization.PublicFormat.SubjectPublicKeyInfo
    )

    # Print the PEM-formatted public key
    public_key = public_key_pem.decode()
    
    
    if public_key:
        try:
            # Decode and verify the token using the public key
            decoded_token = jwt.decode(token, public_key, algorithms=['RS256'])
            return decoded_token
        except jwt.ExpiredSignatureError:
            print('Token has expired.')
        except jwt.InvalidTokenError as e:
            print('Invalid token.' + str(e))
    else:
        print('Failed to retrieve public key for token validation.')
    return None

# Validate the bearer token
decoded_token = validate_bearer_token(token)
if decoded_token:
    print('Token is valid.')
    # You can access the claims in the decoded token using decoded_token['claim_name']
else:
    print('Token is invalid.')
```
